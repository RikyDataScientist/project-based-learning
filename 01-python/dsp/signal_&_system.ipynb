# Import Depedencies

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Generate and Plot Simple Signals

```python
t = np.linspace(-1, 1, 500)
frequency = 5
sine_wave = np.sin(2 * np.pi * frequency * t)
square_wave = np.sign(sine_wave)

plt.figure(figsize=(12, 4))
plt.plot(t, sine_wave, color='red', label='Sine Wave (5 Hz)')
plt.plot(t, square_wave, linestyle='--', label='Square Wave')

ax = plt.gca()
ax.spines[['top', 'right']].set_visible(False)

plt.grid(axis='y', linestyle='--', alpha=0.25)
plt.title('Basic Signal', fontsize=14, weight='bold')
plt.xlabel('Time (S)')
plt.legend()
plt.tight_layout()
plt.show()
```

# Operation

## Basix Operation

### Addition

```python
t = np.linspace(-1, 1, 500)
sine_wave_1 = np.sin(2 * np.pi * 5 * t)
sine_wave_2 = np.cos(2 * np.pi * 7.5 * t)
y = sine_wave_1 + sine_wave_2

plt.figure(figsize=(12, 4))
plt.plot(t, sine_wave_1, color='blue', label='First Wave')
plt.plot(t, sine_wave_2, color='green', label='Second Wave')
plt.plot(t, y, color='red', label='Addition Result')

ax = plt.gca()
ax.spines[['top', 'right']].set_visible(False)

plt.grid(axis='y', linestyle='--', alpha=0.25)
plt.title('Addition Operation', fontsize=14, weight='bold')
plt.xlabel('Time (S)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Amplitudo Scaling

```python
t = np.arange(0, 0.03, 0.001)
x = np.sin(2 * np.pi * 13 * t)
scaled = 2 * x

plt.figure(figsize=(12, 4))
plt.stem(t, x, label='Normal Wave')
plt.plot(t, scaled, linestyle='--', color='green', label='Scaling Wave')

ax = plt.gca()
ax.spines[['top', 'right']].set_visible(False)

plt.grid(axis='y', linestyle='--', alpha=0.25)
plt.title('Amplitudo Scaling', fontsize=14, weight='bold')
plt.xlabel('Time (s)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Time Shift

```python
t = np.linspace(-1, 1, 500)
frequency = 7
sine_wave3 = np.sin(2 * np.pi * frequency * t)
shift = np.roll(sine_wave3, -100)

plt.figure(figsize=(12, 4))
plt.plot(t, sine_wave3, color='red', label='Sine Wave (5 Hz)')
plt.plot(t, shift, linestyle='--', label='Time Shifted Signal')

ax = plt.gca()
ax.spines[['top', 'right']].set_visible(False)

plt.grid(axis='y', linestyle='--', alpha=0.25)
plt.title('Time Shift', fontsize=14, weight='bold')
plt.xlabel('Time (S)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Convolution

Convolution is the process of combining two signals to produce a new signal.

```python
x = [1, 2, 3]
h = [0, 1, 0.5]

# Manual Convolution
def manual_convolution(x, h):
    y = np.zeros(len(x) + len(h) - 1)
    for i in range(len(x)):
        for j in range(len(h)):
            y[i + j] += x[i] * h[j]
    return y
print('Manual Convolution Result:', manual_convolution(x, h))

# Numpy Convolution
y = np.convolve(x, h)
print('Numpy Convolution Result:', y)
```
